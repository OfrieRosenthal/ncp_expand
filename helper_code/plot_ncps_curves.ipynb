```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file1 = ("../target/ncp.facebook_random.tab", "random")
file2 = ("../target/ncp.facebook_wncp.tab", "wncp")
file3 = ("../target/ncp.facebook_combined.tab", "original")
```

```python
# Load the files (skip comment lines starting with '#')
df1 = pd.read_csv(file1[0], sep="\t", comment="#", header=None, names=["k", "phi"])
df2 = pd.read_csv(file2[0], sep="\t", comment="#", header=None, names=["k", "phi"])
df3 = pd.read_csv(file3[0], sep="\t", comment="#", header=None, names=["k", "phi"])

# Plot
plt.figure(figsize=(8, 5))
plt.plot(df1["k"], df1["phi"], marker='o', label=f"{file1[1]}", linewidth=0.8, markersize=1)
plt.plot(df2["k"], df2["phi"], marker='s', label=f"{file2[1]}", linewidth=0.8, markersize=1)
plt.plot(df3["k"], df3["phi"], marker='^', label=f"{file3[1]}", linewidth=0.8, markersize=1)  # added file3

# Logarithmic axes
plt.xscale('log')
plt.yscale('log')

# Axis control (avoid zero, add padding)
plt.xlim(1, max(df1["k"].max(), df2["k"].max(), df3["k"].max()) * 1.1)
plt.ylim(min(df1["phi"].min(), df2["phi"].min(), df3["phi"].min()) * 0.8, 
         max(df1["phi"].max(), df2["phi"].max(), df3["phi"].max()) * 1.2)

plt.xlabel("Cluster size (k, log scale)")
plt.ylabel("Conductance / Phi (log scale)")
plt.title("NCP Comparison (Log-Log Scale)")
plt.legend()
plt.grid(True, which="both", linewidth=0.5)
plt.tight_layout()
plt.show()
```
